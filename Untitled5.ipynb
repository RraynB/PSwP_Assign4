```python
# pairing strategy, opponent will be random of the leftover places in the matrix
# even n, opposition goes first
import numpy as np
def game(n):

    def pairing_strategy(n,pos):   
        if pos[1] % 2 == 0:
            new_col = pos[1] + 1
        else: 
            new_col = pos[1] - 1
    
        return pos[0],new_col
        
    def random_strategy(empty_positions): 
        return tuple(empty_positions[np.random.randint(len(empty_positions))])


    matrix = np.full((n, n), -1)
    empty_positions = np.argwhere(matrix == -1)
    while len(empty_positions) != 0: 
        opp_pos = random_strategy(empty_positions)
        matrix[opp_pos] = 1
        matrix[pairing_strategy(n,opp_pos)]=0
        empty_positions = np.argwhere(matrix == -1)

    return int(np.linalg.det(matrix))

for i in range(50): 
    print(game(4))
```
